```python
import duckdb
```

```python
conexao = duckdb.connect(database='dados_duckdb.db', read_only=False)
```

```python
df = conexao.execute("SELECT * FROM produtos").fetchdf()
df.head(10)
```

```python
df2 = df.drop(columns=['id_categoria', 'id_fornecedor'])
df2.head(10)
```

```python
#criação de tabela da camada Gold (dim_produtos)
conexao.execute("""
CREATE TABLE IF NOT EXISTS dim_produtos (
                id_produto BIGINT,
                nm_produto TEXT,
                vl_produto FLOAT
                )
""")
```

```python
#inserção de dados na tabela da camada Gold (dim_produtos)
conexao.execute("INSERT INTO dim_produtos SELECT * FROM df2")
```

```python
df_dim = conexao.execute("SELECT * FROM dim_produtos").fetchdf()
df_dim.head(10)
```

```python
conexao.close()
```
