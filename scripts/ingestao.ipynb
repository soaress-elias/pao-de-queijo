```python
import duckdb
import pandas as pd
import os
from datetime import datetime
```

```python
conexao = duckdb.connect(database='dados_duckdb.db', read_only=False)
```

```python
arquivo1 = 'z0019_2.csv'
data_ingestao = datetime.now()
df = pd.read_csv(f'../landing/{arquivo1}', sep=';')
df['nome_arquivo'] = arquivo1
df['data_ingestao'] = data_ingestao
df.head()
```

```python
#criação de tabela da camada Bronze
conexao.execute("""
CREATE TABLE IF NOT EXISTS bronze_z0019 (
                NATBR VARCHAR,
                MAKTX VARCHAR,
                WERKS VARCHAR,
                MAINS VARCHAR,
                LABST VARCHAR,
                nome_arquivo VARCHAR,
                data_ingestao TIMESTAMP
                )
""")
```

```python
#ingestao na tabela da camada Bronze (bronze_z0019)
conexao.execute("INSERT INTO bronze_z0019 SELECT * FROM df")
```

```python
resultado1 = conexao.execute("SELECT * FROM bronze_z0019").fetchdf()
resultado1.head()
```
